```python
  import pandas as pd
  import numpy as np
  import seaborn as sns
  import matplotlib.pyplot as plt
   
  datos = {
      "nota":           [16, 10, 18, 12, 14, 9, 17, 11],
      "horas_estudio":  [7, 3, 9, 5, 6, 2, 8, 4],
      "horas_series":   [2, 6, 1, 5, 3, 7, 2, 5],
      "altura_cm":      [172, 178, 165, 180, 175, 163, 169, 158],
  }
  df = pd.DataFrame(datos)
  corr = df.corr()
  print(corr.round(4))
```

```python
  fig, ax = plt.subplots(figsize=(6, 5))
  sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
              vmin=-1, vmax=1, square=True,
              cbar_kws={"label": "Correlacion de Pearson"}, ax=ax)
  plt.title("Matriz de correlacion - seaborn")
  plt.tight_layout()
  plt.savefig("heatmap_seaborn_completo.png")
```

```python
  mascara = np.triu(np.ones_like(corr, dtype=bool))
  print("Celdas visibles:", (~mascara).sum(), "de", mascara.size)
   
  fig, ax = plt.subplots(figsize=(6, 5))
  sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
              vmin=-1, vmax=1, square=True, mask=mascara,
              cbar_kws={"label": "Correlacion de Pearson"}, ax=ax)
  plt.title("Matriz de correlacion - seaborn (con mascara)")
  plt.tight_layout()
  plt.savefig("heatmap_seaborn_mascara.png")
```
