```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Completa: reconstruye tu dataset del Tema 9 (Calidad Ambiental en Cusco)
datos = {
    "estacion":  ["Centro", "San Jerónimo", "Wanchaq", "San Sebastián", "Poroy"],
    "pm10_ugm3": [45.0, 28.0, 62.0, 35.0, np.nan],
    "no2_ppb":   [18.2, 12.5, 25.4, 15.0, 9.8],
    "ruido_db":  [72, 58, 81, 65, 52]
}
df = pd.DataFrame(datos).set_index("estacion")

# Imputamos el NaN de Poroy con la mediana (40.0) como en el Tema 9 y calculamos la correlacion
df_limpio = df.fillna(df.median())
corr = df_limpio.corr()
print(corr.round(4))
```

```python
# Completa: heatmap con sns.heatmap()
plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=1
)
plt.title("Matriz de correlacion ambiental con seaborn")
plt.tight_layout()
plt.savefig("heatmap_seaborn.png", dpi=150, bbox_inches="tight")
plt.show()
print("Guardado como heatmap_seaborn.png")
```

```python
# Completa: mascara y segundo heatmap
mascara = np.triu(np.ones_like(corr, dtype=bool))
print("Celdas visibles:", int((~mascara).sum()), "de", corr.size)

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    mask=mascara,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=1
)
plt.title("Matriz de correlacion con mascara triangular")
plt.tight_layout()
plt.savefig("heatmap_seaborn_mascara.png", dpi=150, bbox_inches="tight")
plt.show()
print("Guardado como heatmap_seaborn_mascara.png")
```
