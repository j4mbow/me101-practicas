```python
import pandas as pd
import numpy as np

# Completa: crea el diccionario con al menos 6 estudiantes y al menos 2 np.nan
datos = {
    "nombre": ["Abel", "Luana", "Salvador", "Tania", "Carlos", "Lucia"],
    "curso": ["ME101", "ME101", "ME102", "ME102", "ME101", "ME102"],
    "nota": [16, 12, 18, np.nan, 8, 14],
    "asistencia_pct": [80, 90, np.nan, 75, 60, 85]
}

df_original = pd.DataFrame(datos)
# Completa: guarda como CSV
df_original.to_csv("estudiantes.csv", index=False)

print(df_original)
print("Archivo estudiantes.csv creado.")
```

```python
# Completa: lee el CSV
df = pd.read_csv("estudiantes.csv")
# Completa: muestra shape y cantidad de NaN por columna
print("Dimensiones:", df.shape)
print("Valores faltantes por columna:")
print(df.isna().sum())
```

```python
# Completa: elimina filas con nota faltante
df_limpio = df.dropna(subset=["nota"]).copy()

# Completa: rellena asistencia_pct faltante con el promedio
media_asistencia = df_limpio["asistencia_pct"].mean()

# Completa: verifica que no queden valores faltantes
df_limpio["asistencia_pct"] = (
    df_limpio["asistencia_pct"].fillna(media_asistencia)
)

# Mostrar las comprobaciones.
print("Media usada para imputar:", media_asistencia)
print("Dimensiones después de limpiar:", df_limpio.shape)

print("\nValores faltantes después de limpiar:")
print(df_limpio.isna().sum())

print("\nDatos limpios:")
print(df_limpio)
```

```python
# Completa: promedio de nota agrupado por curso
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()

print("Promedio de nota por curso:")
print(promedio_por_curso)

df_limpio.to_csv("estudiantes_limpio.csv", index=False)

print("Archivo estudiantes_limpio.csv guardado.")
```
