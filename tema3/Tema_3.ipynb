```python
import numpy as np
# Completa: crea el array 'notas' con los 6 valores indicados
notas = np.array([11, 14, 20, 8, 16, 13])
# Completa: imprime shape, dtype y ndim
print(notas.shape)
print(notas.dtype)
print(notas.ndim)
```

```python
# Completa: promedio, maximo, y cantidad de aprobados
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 10.5).sum() # pista: suma un array booleano
print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd
# Completa: crea el diccionario de datos y el DataFrame
datos = {
"nombre": ["Abel", "Luana", "Salvador", "Tania", "Carlos"],
"nota": [2, 13, 12, 17, 6],
"asistencia_pct": [10, 66, 48, 80, 30],
}
df = pd.DataFrame(datos)
# Completa: filtra estudiantes con nota >= 10.5 Y asistencia_pct >= 70
# pista: combina dos condiciones booleanas con el operador &
aprobados = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]
print(aprobados)
```
