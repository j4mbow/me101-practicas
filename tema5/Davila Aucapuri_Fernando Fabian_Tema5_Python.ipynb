```python
import pandas as pd
import numpy as np
from scipy import stats
df = pd.read_csv("estudiantes_limpio.csv") # generado en el Tema 4
# Completa: media, mediana, std muestral (pandas), var poblacional (numpy), Q1, Q3, IQR
media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()
var_poblacional = np.var(df["nota"].to_numpy())
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print(f"\nMedia: {media:.4f}")
print(f"Mediana: {mediana:.4f}")
print(f"Desviacion estandar muestral: {std_muestral:.4f}")
print(f"Varianza poblacional: {var_poblacional:.4f}")
print(f"Q1: {q1:.4f}")
print(f"Q3: {q3:.4f}")
print(f"IQR: {iqr:.4f}")
```

```python
# Completa: skewness y kurtosis con pandas Y con scipy
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()

skew_scipy = stats.skew(df["nota"])
kurt_scipy = stats.kurtosis(df["nota"])

print(f"Skewness pandas: {skew_pandas:.4f}")
print(f"Kurtosis pandas: {kurt_pandas:.4f}")
print(f"Skewness scipy: {skew_scipy:.4f}")
print(f"Kurtosis scipy: {kurt_scipy:.4f}")
```

```python
# Completa: define la funcion resumen_estadistico
def resumen_estadistico(serie, decimales=4):
    media = float(serie.mean())
    desv_std = float(serie.std())

    return {
        "n": len(serie),
        "media": round(media, decimales),
        "mediana": round(float(serie.median()), decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round((desv_std / media) * 100, decimales)
    }

print(resumen_estadistico(df["nota"]))
```

```python
# Completa: define clasificar_dispersion con if/elif/else
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(resumen["cv_pct"])

    print(
        f"{columna}: media={resumen['media']:.4f}, "
        f"desv_std={resumen['desv_std']:.4f}, "
        f"CV={resumen['cv_pct']:.4f}% -> {nivel}"
    )
```
